Slicing

```python
import numpy as np
```

```python
a = np.array([1,2,3,4,5,6,7,8])
```

```python
print(a[:3])
```

```python
print(a[:3:2])
```

```python
print(a[::2]) # Lower limit defaults to 0, Upper limit is the max which is index 9 as there are 9 elements and then step size of two
```

```python
print(a[-3:]) #Last three numbers
```

```python
# OR positive index
print(a[5:])
```

```python
# Matrix
```

```python
b = np.arange(1, 13).reshape(3,4)
print(b)
```

```python
b[1:3, 1:3]
```

```python
b[1:, 1:3] # Takes row 1 till max row - 1, column 1 till 3-1
```

```python
b[1:, 1:3]
```

```python
b[1:,2:]
```

```python
b[2,1:3]
```

```python
b[2,1]
```

```python
b[2]
```

```python
b[-1]
```

```python
b[:, -1]
```

```python
# Tensor
```

```python
c = np.arange(1,37).reshape(3,3,4)
```

```python
print(c)
```

```python
print(c[1,1,:])
```

```python
print(c[0]) # same as c[0,:, :] or c[0,:]
```

```python
print(c[2, 0:2,1:3])
```

```python
d = np.arange(3*4*5).reshape(3,4,5)
print(d)
```

```python
print(d[0,::2,:]) # First Matrix, then row 0, row 2. and then all columns
```

```python
print(d[2,:,::2])
```

```python
print(d[2,:,2::2])
```

```python
import pandas as pd
import numpy as np
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
nltk.download('all')
```

```python
# Sample Data

data = [
    {'message': 'i like the movie', 'sentiment': 'positive'},
    {'message': 'movie is not good', 'sentiment': 'negative'},
    {'message': 'wasted my time', 'sentiment': 'negative'},
    {'message': 'liked the movie', 'sentiment': 'positive'}
]
```

```python
df = pd.DataFrame(data)
df
```

```python
stop_Words = set(stopwords.words('english'))
stop_Words
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
# Convert text data into numerical. Build Corpus
```

```python
def process_text(text):
  text = text.lower()
  text = ''.join([char for char in text if char not in string.punctuation])
  tokens = word_tokenize(text)
  tokens = [word for word in tokens if word.isalpha()]
  tokens = [word for word in tokens if word not in stop_Words]
  tokens = [lemmatizer.lemmatize(word) for word in tokens]
  return ' '.join(tokens)
```

```python
df['tokens'] = df['message'].apply(process_text)
```

```python
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorize = CountVectorizer() # Build the Corpus
X = vectorize.fit_transform(df['tokens'])
print(vectorize.get_feature_names_out())
print(X.toarray())
```

```python
y = df['sentiment'].map({'positive': 1, 'negative': 0})
y
```

```python
model = LogisticRegression()
model.fit(X, y)
model.score(X, y)
```

```python
test_message = 'movie is not good'
test_message = process_text(test_message)
test_message
```

```python
test_data = vectorize.transform([test_message])
test_data
```

```python
model.predict(test_data)
```

```python
positive_message = 'liked the movie'
positive_message = process_text(positive_message)
positive_message
```

```python
positive_test_data = vectorize.transform([positive_message])
positive_test_data
```

```python
model.predict(positive_test_data)
```

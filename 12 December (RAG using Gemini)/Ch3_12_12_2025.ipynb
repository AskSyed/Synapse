```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_community.document_loaders import PyPDFLoader
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
GOOGLE_API_KEY = "AIzaSyAB_tQ2biQ7RmNok47-LU9zh4bT_M_3naw"
```

```python
# Step 1 Load
pdf_path = "/content/doc202517482201.pdf"
loader = PyPDFLoader(pdf_path)
pages = loader.load()
print("Number of Pages", len(pages))
```

```python
# Split
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100)
```

```python
chunks = splitter.split_documents(pages)
print("Number of Chunks", len(chunks))
```

```python
# Step 2 Created Embeddings (Numerical format of text)
embeddings = GoogleGenerativeAIEmbeddings(
    model="text-embedding-004", google_api_key=GOOGLE_API_KEY
)
```

```python
vectorstore = FAISS.from_documents(chunks, embeddings)
```

```python
print("Gemini Embeddings Created")
```

```python
query = "What is the GDP growth rate of India for 2024-25?"
```

```python
# Step 3
result = vectorstore.similarity_search(query, k=3)
```

```python
context = "\n\n".join([doc.page_content for doc in result])
```

```python
# Step 4
prompt = f"""
Use the following document context to answer the question.
context: {context}
---
question: {query}
---
Answer :
"""
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", google_api_key=GOOGLE_API_KEY)

rag_response = llm.invoke(input=prompt)
print(rag_response.content.strip())
```

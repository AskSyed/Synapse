```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
GOOGLE_API_KEY = "AI**********************************w"
```

```python
#step1
pdf_path = "/content/doc202517482201.pdf"
loader = PyPDFLoader(pdf_path)
pages = loader.load()
print("Number of Pages ",len(pages))
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100,

)
```

```python
chunks = splitter.split_documents(pages)
print("Number of chunks ",len(chunks))
```

```python
#step 2
embeddings = GoogleGenerativeAIEmbeddings(model="text-embedding-004",google_api_key=GOOGLE_API_KEY)
```

```python
vectorstore = FAISS.from_documents(chunks, embeddings)
```

```python
print("Gemini Embeddings Created")
```

```python
query = "What is the GDP growth rate of india for 2024-25?"
```

```python
#step3
results= vectorstore.similarity_search(query,k=3)
```

```python
context = "\n\n".join([doc.page_content for doc in results])
```

```python
#step 4
prompt=f"""
use the following document context to answer the question.
context:{context} question:{query} Answer :"""
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash",google_api_key=GOOGLE_API_KEY)
rag_response = llm.invoke(input=prompt)
print(rag_response.content.strip())
```

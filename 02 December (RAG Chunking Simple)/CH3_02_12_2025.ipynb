```python
import re
from typing import List
```

```python
paragraph = '''Artificial intelligence is transforming healthcare rapidly. Machine learning models can now detect desease from medical images with remarkable accuracy. This technology promises to make diagnosis faster and more accessible worldwide.'''
```

```python
print(paragraph)
```

```python
# Paragraph based Chunking
# Start with Python but libraries can be used (such as NLP)

def paragraph_chunking(text: str) -> List[str]:
  paragraphs = [p.strip() for p in text.split('\n\n') if p.strip()]
  return paragraphs
```

```python
paragrahs = paragraph_chunking(paragraph)
print(paragrahs)
```

```python
# Step 2
# Sentence based chunking

def sentence_tokenizer(text: str) -> List[str]:
  sentences = re.split(r'(?<=[.!])\s+', text.strip())
  return [s.strip() for s in sentences if s.strip()]

#
```

```python
sentences = sentence_tokenizer(paragraph)
for i, chunk in enumerate(sentences):
  print(f'Chunk {i + 1}: {chunk}')
```

```python
# Fixed size token
def fixed_size(text: str, chunk_size: int=10) -> List[str]:
  words = text.split()
  chunks = []
  for i in range(0, len(words), chunk_size):
    chunk = ' '.join(words[i:i+chunk_size])
    chunks.append(chunk)
  return chunks
#

def fixed_size_char_chunking(text: str, chunk_size: int = 10) -> list[str]:
  return [text[i:i+chunk_size] for i in range(0, len(text), chunk_size)]
```

```python
chunks = fixed_size_char_chunking(paragraph)
for i, chunk in enumerate(chunks):
  print(f'Chunk {i + 1}: {chunk}')
```

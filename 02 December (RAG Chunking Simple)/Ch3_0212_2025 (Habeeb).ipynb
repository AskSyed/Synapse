```python
p = '''Artificial intellegence is transforming healthcare rapidly.Machine learning models can now detect desease from medical images with remarkable accuray.This technology promises to make diagnosis faster and more accesible worldwide.'''
```

```python
print(p)
```

```python
import re
from typing import List
```

```python
#Paragraphbased Chunking
def paragraph_chunking(text:str)->List[str]:
  paragraphs = [p.strip() for p in text.split('\n\n') if p.strip()]
  return paragraphs
```

```python
paragraphs = paragraph_chunking(p)
print(paragraphs)
```

```python
#step 2 - Sentence Based Chunking
def sentence_tokenizer(text:str)->List[str]:
  sentences = re.split(r'(?<=[.!])\s+',text.strip())
  return [s.strip() for s in sentences if s.strip()]
```

```python
sentences = sentence_tokenizer(p)
for i,chunk in enumerate(sentences):
  print(f'Chunk {i+1}: {chunk}')
```

```python
Chunk 1: Artificial intelligence is transforming healthcare rapidly.
Chunk 2: Machine learning models can now detect desease from medical images with remarkable accuracy.
Chunk 3: This technology promises to make diagnosis faster and more accessible worldwide.
```

```python
sentences1 = p.split('.')
for i,chunk in enumerate(sentences1):
  print(f'Chunk {i+1}: {chunk}')
```

```python
def fixed_size(text:str,chunk_size:int=10) -> List[str]:
  chunks = []
  for i in range(0,len(text),chunk_size):
    chunk = text[i:i+chunk_size]
    chunks.append(chunk)
  return chunks
```

```python
chunks = fixed_size(p,chunk_size=10)
for i ,chunk in enumerate(chunks):
  print(f'Chunk {i+1}: {chunk}')
```

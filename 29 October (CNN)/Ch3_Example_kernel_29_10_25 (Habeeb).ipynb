```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('/content/dog.png',cv2.IMREAD_GRAYSCALE)
```

```python
kernel = np.array([
    [-2,-1,0],
    [-1,1,1],
    [0,1,2]
])
```

```python
edges = cv2.filter2D(img,-1,kernel=kernel)
```

```python
plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(edges,cmap='gray')
plt.axis('off')
plt.show()
```

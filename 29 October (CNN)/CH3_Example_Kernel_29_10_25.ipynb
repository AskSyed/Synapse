```python
import numpy as np # to create array 2 dimensional (Matrix)
import cv2 # To read image - OpenCV python
import matplotlib.pyplot as plt
import os
```

```python
img = cv2.imread('/content/beagle-hound-dog.png', cv2.IMREAD_GRAYSCALE)
```

```python
kernel = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1],
])
```

```python
# Sharpen filter

kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0],
])
```

```python
# CV2 is to import OpenCV image processing library
edges = cv2.filter2D(img, -1, kernel)
```

```python
# matplot lib is to draw
plt.figure(figsize=(10, 10))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title('Sharpen')
plt.axis('off')

plt.tight_layout()
plt.show()
```
